```python
import warnings
warnings.filterwarnings('ignore')
```

```python
from langchain_community.vectorstores import Chroma
from langchain_core.documents import Document
```

```python
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
os.getenv('hf_emb')
```

```python
doc1 = Document(
    page_content='this is fang yuan, morality is weakness',
    metadata={'name':'reverend insanity','author':'ren zhuo'}
)
doc2 = Document(
    page_content='ayanokouji kiyotaka, people are nothing but tools',
    metadata={'name':'COTE','author':'kiyopyon'}
)
doc3 = Document(
    page_content='naruto uzumaki never gives up, has byakugan clove',
    metadata={'name':'naruto','author':'kun'}
)
doc4 = Document(
    page_content='ore wa monkey d luffy, kaizouku ni naru otoko da, zoro daiski, nami bazooka skiii, mecha mecha ero skiii',
    metadata={'name':'one piece','author':'oda'}
)
```

```python
from langchain_huggingface.embeddings import HuggingFaceEndpointEmbeddings
```

```python
emb = HuggingFaceEndpointEmbeddings(repo_id=os.getenv('hf_emb'))
```

```python
emb.embed_query('something')
```

```python
emb.embed_query('whatever love')
```

```python
emb.embed_documents([doc1.page_content])
```

```python
vectorstore = Chroma(
    embedding_function=emb,
    collection_name='anime_stuff',
    persist_directory='chroma_db'
)
```

```python
docs = [doc1, doc2, doc3, doc4]
```

```python
vectorstore.add_documents(docs)
```

```python
vectorstore.get(include=['embeddings','documents','metadatas'])
```

```python
doc5 = Document(
    page_content='hero villan commoner magic elements',
    metadata={'name':'s','author':'me'}
)
```

```python
vectorstore.add_documents([doc5])
```

```python
result = vectorstore.similarity_search(
    query='dark flame master',
    k=2
)
for doc in result:
    print(doc.page_content)
```

```python
result = vectorstore.similarity_search(
    query='china',
    k=2
)
for doc in result:
    print(doc.page_content)
```

```python
retriever = vectorstore.as_retriever(search_kwargs={'k':2})
```

```python
retriever2 = vectorstore.as_retriever(search_type='mmr',search_kwargs={'k':2,'lambda_mul':1})
```

```python
retriever.invoke('classroom of the elite')
```
